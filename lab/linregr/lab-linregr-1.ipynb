# Лінійна регресія: теоретичний вступ

## Для чого потрібна лінійна регресія?

Лінійна регресія — один з найпростіших і найпоширеніших статистичних методів, який дозволяє описати **лінійний зв’язок** між двома кількісними змінними:

- **незалежною змінною** \( x \) (predictor, explanatory variable)
- **залежною змінною** \( y \) (response, outcome)

У біології це дуже часто потрібно, коли ми хочемо зрозуміти, як одна величина впливає на іншу. Приклади:

- залежність швидкості росту бактерій від концентрації поживних речовин;
- зв’язок між масою тіла тварини і її метаболічним рівнем;
- вплив дози препарату на відсоток виживання клітин;
- залежність кількості хлорофілу від інтенсивності освітлення тощо.

Метод шукає **пряму лінію**, яка найкраще описує цю залежність у вигляді рівняння:

$$
y = \text{slope} \cdot x + \text{intercept}
$$

де:
- `slope` — кутовий коефіцієнт (наскільки змінюється \( y \) при збільшенні \( x \) на 1 одиницю);
- `intercept` — точка перетину з віссю \( y \) (значення \( y \), коли \( x = 0 \)).

---

## На які питання дозволяє відповісти лінійна регресія?

1. **Чи існує лінійний зв’язок** між \( x \) і \( y \)?
2. **Наскільки сильний** цей зв’язок? (оцінюємо за \( $R^2$ \))
3. **Який напрямок** зв’язку? (позитивний чи негативний — за знаком `slope`)
4. **Наскільки точно** ми можемо передбачати значення \( y \) за відомим \( x \)?
5. **Чи адекватна** отримана модель даним? (аналіз залишків)

---

## Що ми отримуємо на виході?

Після виконання `stats.linregress(x, y)` ми отримуємо об’єкт, з якого найчастіше беремо:

| Параметр     | Що означає                                      | Як інтерпретувати |
|--------------|--------------------------------------------------|-------------------|
| `slope`      | Кутовий коефіцієнт                               | Зміна \( y \) при збільшенні \( x \) на 1 |
| `intercept`  | Вільний член                                     | Значення \( y \) при \( x = 0 \) |
| `rvalue`     | Коефіцієнт кореляції Пірсона \( r \)             | Від −1 до +1 |
| \( $R^2$ \)    | Коефіцієнт детермінації (`rvalue**2`)            | Частка варіації \( y \), яку пояснює модель (від 0 до 1) |
| `pvalue`     | p-значення для перевірки \( $H_0: \text{slope} = 0$ \) | Чи статистично значущий зв’язок |
| `stderr`     | Стандартна помилка оцінки `slope`                | Точність оцінки нахилу |


## Як оцінити адекватність моделі?

Навіть якщо \( $R^2$ \) високий, модель може бути поганою. Тому обов’язково перевіряємо:

### 1. Коефіцієнт детермінації \( $R^2$ \)
- \( $R^2$ \approx 1 \) — модель добре пояснює дані  
- \( $R^2$ \approx 0 \) — лінійний зв’язок майже відсутній  
- У біології часто задовільним вважають \( $R^2$ > 0.6–0.7 \), але все залежить від контексту.

### 2. Аналіз залишків (residuals)
Залишки = різниця між реальними і передбаченими значеннями:
$$
\text{residuals} = y - y_{\text{pred}}
$$

**Добре**, якщо:
- залишки випадково розкидані навколо нуля;
- немає очевидних трендів, дуг, «воронки» (гетероскедастичності);
- немає сильних викидів.

**Погано**, якщо:
- залишки утворюють дугу → зв’язок нелінійний;
- розкид збільшується з ростом \($ y_{\text{pred}} $\) → порушена гомоскедастичність;
- є систематичний зсув вгору/вниз.

Саме тому в шаблоні є окремий блок «Аналіз залишків» з графіком `residuals ~ y_pred`.

### 3. Візуальна перевірка

- scatter-plot експериментальних точок + лінію регресії;
- графік залишків.

## Умови застосування методу (припущення)

Лінійна регресія коректна лише за певних умов:

1. **Лінійність** — справжній зв’язок між \( x \) і \( y \) близький до прямої.
2. **Незалежність спостережень** — кожна точка вимірювалася незалежно.
3. **Гомоскедастичність** — дисперсія залишків приблизно однакова на всьому діапазоні \( x \).
4. **Нормальність залишків** (бажано) — для коректних p-значень і довірчих інтервалів.
5. **Відсутність сильних викидів**, які непропорційно впливають на нахил.

Якщо умови порушені — модель може давати хибні висновки. Тоді варто розглянути трансформації даних (логарифм, квадратний корінь) або нелінійні моделі.

## Типові помилки інтерпретації

Навіть добре побудована регресія легко може бути неправильно зрозуміла. Найчастіші помилки:

1. **Високий \( $R^2$ \) ≠ причинно-наслідковий зв’язок**  
   Лінійна регресія показує лише **асоціацію**, а не те, що \( x \) **спричиняє** зміну \( y \).  
   Приклад: висока кореляція між кількістю лелек і народжуваністю не означає, що лелеки приносять дітей.

2. **Екстраполяція за межі даних**  
   Модель надійна лише в діапазоні значень \( x \), на яких вона була побудована.  
   Передбачення для дуже малих або дуже великих \( x \) часто буває абсолютно хибним.

3. **Ігнорування біологічного сенсу intercept**  
   Якщо \( x = 0 \) неможливе (наприклад, нульова концентрація субстрату, нульова маса тіла), то значення `intercept` не варто інтерпретувати буквально.

4. **Один викид може «зламати» нахил**  
   Навіть одна аномальна точка сильно впливає на `slope` і \( $R^2$ \). Завжди перевіряйте графік залишків і наявність outliers.

5. **\( R^2 \) сам по собі не говорить про статистичну значущість**  
   При великій вибірці навіть слабкий зв’язок може дати високий \( $R^2$ \) і дуже мале p-value. Завжди дивіться і на p-value, і на розмір ефекту (`slope`).

6. **Порівняння \( R^2 \) між різними моделями з різною кількістю точок**  
   \( $R^2$ \) чутливий до обсягу вибірки. Краще порівнювати моделі на однакових даних.

| **`scipy.stats.linregress`** | Дуже простий синтаксис<br>Одразу дає `slope`, `intercept`, \( r \), \( $R^2$ \), p-value, stderr | Працює лише з **однією** незалежною змінною | Навчання, швидкий аналіз, біологічні задачі з 1 предиктором |

**Висновок для цього курсу:**  
`linregress` — ідеальний інструмент для першого знайомства і більшості біологічних задач з однією незалежною змінною. Він швидко дає все найважливіше і не відволікає зайвим кодом.

## Як інтерпретувати \( $R^2$ \) у біологічному контексті

\( $R^2$ \) показує, **яку частку варіації** залежної змінної пояснює модель. У біології «добрий» \( $R^2$ \) сильно залежить від системи:


| Діапазон $R^2$ | Сила зв'язку / Ефект | Рекомендації щодо $P$-value | Приклади з біологічної практики | Інтерпретація контексту |
|---|---|---|---|---|
| 0.80 – 1.00 | Дуже сильний / винятковий зв'язок | Обов'язково $p < 0.001$. Якщо $p$ велике, це математичний артефакт малої вибірці. | Лабораторна калібрувальна крива концентрації речовини; вимірювання оптичної густини білка. | Рідкість для дикої природи. Свідчить про жорсткий лабораторний контроль, де усунуто майже весь сторонній шум. |
| 0.50 – 0.79 | Сильний зв'язок (Strong effect) | Очікується $p < 0.01$ або $p < 0.001$. Сигнал чіткий і легко підтверджується. | Фізіологічні виміри в ізольованих умовах (напр., вплив дози препарату на культуру клітин in vitro). | Чудовий і переконливий результат для біології. Фактор є домінуючим у процесі. |
| 0.30 – 0.49 | Помірний / середній зв'язок (Moderate effect) | Потрібно $p < 0.05$, бажано $p < 0.01$ для впевненості, що це не випадковість. | Залежність маси тварини від калорійності раціону; динаміка росту рослин від середньої температури. | Модель вловлює чіткий біологічний тренд, хоча більше половини варіації зумовлено іншими прихованими факторами. |
| 0.10 – 0.29 | Слабкий зв'язок (Low effect) | Критично мати $p < 0.05$ (на великій вибірці). Якщо $p > 0.05$ — модель повністю неціннісна. | Польова екологія (напр., вплив вологості на чисельність комах); генетичні асоціації (вплив одного гена на ознаку). | Сигнал слабкий, але біологічно реальний. Фактор працює в комплексі з десятками інших складних умов. |
| 0.00 – 0.09 | Ефект відсутній або зневажливо малий | Зазвичай $p > 0.05$ (статистично незначуще). Навіть якщо $p < 0.05$ через гігантську вибірку, ефект мікроскопічний. | Спроба пов'язати колір очей ссавця з його швидкістю бігу; випадковий шум. | Модель не має практичної чи прогнозувальної сили. Шум повністю заблокував або знищив "сигнал". |


**Важливо:**  
У контрольованих лабораторних експериментах ми очікуємо вищі \( $R^2$ \), ніж у польових екологічних дослідженнях. Низький \( $R^2$ \) не завжди означає «погану» модель — іноді це просто відображення складності біологічної системи.

------------------------------
## 1. Аналогія $R^2$ з співвідношенням "сигнал/шум"
Математично загальна мінливість (дисперсія) нашої залежної змінної $Y$ ділиться на дві частини:  
$\text{Загальна варіативність } (Y) = \text{Варіативність моделі (Сигнал)} + \text{Варіативність залишків (Шум)}$  
Формула $R^2$ виглядає так: 
$$R^2 = \frac{\text{Пояснена моделлю дисперсія (Сигнал)}}{\text{Загальна дисперсія (Сигнал + Шум)}}$$

* $R^2 = 1$: Шуму взагалі немає. Модель ідеально (на 100%) описує кожну точку.
* $R^2 = 0$: Сигналу немає взагалі, або ми не змогли його вловити. Навколо лише білий шум.
* $R^2 = 0.40$: Наш "сигнал" пояснює 40% усієї мінливості даних, а 60% — це випадковий шум або вплив інших факторів, які ми не врахували.

Низький $R^2$ не означає, що залежності між $X$ та $Y$ фізично чи біологічно не існує.
Це може означати дві речі:

   1. Шум занадто сильний: Наприклад, ми досліджуємо вплив добрива ($X$) на врожайність кукурудзи ($Y$). Зв'язок точно є. Але якщо в цей рік була посуха, напад шкідників і град (величезний випадковий шум), то наш $R^2$ буде дуже низьким (наприклад, 0.05). Добриво працювало, але зовнішній шум "заглушив" цей сигнал.
   2. Форма моделі неправильна (хибний приймач сигналу): Якщо зв'язок між $X$ та $Y$ є нелінійним (наприклад, парабола), а ми намагаємося провести пряму лінію, лінійна регресія видасть $R^2 \approx 0$. Сигнал є, але наша лінійна модель "налаштована на іншу частоту".

## 3. Що таке "сила зв'язку" та як інтерпретувати $R^2$?
Сила зв'язку в контексті регресії показує, наскільки міцно зміни однієї змінної асоціюються зі змінами іншої.

* У парній лінійній регресії $R^2$ — це просто квадрат звичайного коефіцієнта кореляції Пірсона ($r$). Якщо $r = 0.7$ (сильна кореляція), то $R^2 = 0.49$.
* Шкала інтерпретації (хоча вона дуже умовна і залежить від галузі):
* $R^2 < 0.2$ — зв'язок дуже слабкий (модель майже нічого не пояснює).
   * $0.2 \le R^2 < 0.5$ — зв'язок слабкий / помірний.
   * $0.5 \le R^2 < 0.7$ — зв'язок середньої сили.
   * $R^2 \ge 0.7$ — зв'язок сильний.

Важливий нюанс щодо галузей:

* В фізиці або інженерії $R^2 = 0.7$ вважається провалом, бо там процеси точні, і очікують $R^2 > 0.95$.
* В соціології, медицині чи економіці $R^2 = 0.25$ (пояснено 25% варіативності) може бути величезним науковим відкриттям, адже поведінка людини чи організму залежить від мільйона факторів, і якщо один ваш фактор пояснює аж чверть — це круто!


* $R^2$ показує практичну значущість (ефект): Наскільки потужний сигнал?
* P-value (p-значення або F-тест) показує статистичну значущість (достовірність): Чи цей сигнал взагалі існує, чи він нам привидився через випадковість?

Парадокс розміру вибірки:

   1. Якщо у вас вибірка 10 000 людей, і ви знайшли слабкий сигнал із $R^2 = 0.01$. Тест скаже: "Цей зв'язок надзвичайно достовірний (p < 0.001), він точно є, це не випадковість!". Але практична цінність моделі нульова — сигнал занадто слабкий, щоб робити прогнози.
   2. Якщо у вас вибірка 3 людини, ви можете отримати $R^2 = 0.99$ (сигнал нібито ідеальний). Але математичний тест скаже: "Недостовірно (p > 0.05), замало даних, це може бути чиста випадковість".

## Алометричні залежності

Багато біологічних співвідношень мають вигляд степеневої функції:

$
y = a \cdot x^{b}
$

Після логарифмічного перетворення отримуємо лінійне рівняння:

$
\log(y) = \log(a) + b \cdot \log(x)
$

Тут **slope** лінійної регресії \($\log(y)$\) від \($\log(x)$\) і є показником степеня \(b\).

Інтерпретація показника \(b\):

| Значення \(b\) | Характер залежності      |
|----------------|--------------------------|
| \($b \approx 1$\) | **Ізометрія**            |
| \(b > 1\)       | **Позитивна алометрія**  |
| \(b < 1\)       | **Негативна алометрія**  |
